```python
#請建立一個範本list
my_list = [1, 2, 3, 4, 5]
#並且建立常用的實體方法
my_list.append(6)  # 在列表末尾添加元素
print(my_list)  # 輸出: [1, 2, 3, 4, 5, 6]
my_list.remove(2)  # 移除列表中的元素2
print(my_list)  # 輸出: [1, 3, 4, 5, 6]
my_list.insert(1, 10)  # 在索引1的位置插入元素10
print(my_list)  # 輸出: [1, 10, 3, 4, 5, 6]
my_list.sort()  # 對列表進行排序
print(my_list)  # 輸出: [1, 3, 4, 5, 6, 10]
my_list.reverse()  # 將列表反轉
print(my_list)  # 輸出: [10, 6, 5, 4, 3, 1] 
my_list.clear()  # 清空列表
print(my_list)  # 輸出: []
```

```python
#我有一個list,要使用for迴圈請將list中的每個元素乘以2
my_list = [1, 2, 3, 4, 5]
for i in my_list:
    print(i * 2)  # 輸出: 2, 4, 6, 8, 10

for word in "APPLE":
    print(word)

```

```python
#請建立國家code與國家名稱的字典
country_dict = {
    "TW": "Taiwan",
    "US": "United States",
    "JP": "Japan",
    "CN": "China",
    "KR": "South Korea"
}
for code in country_dict.keys():
    print(code, country_dict[code])
```

```python
for country in country_dict.values():
    print(country)
    
```

```python
for item in country_dict.items():
    print(item)
```

```python
for key, value in country_dict.items():
    print(key, value)
```
